```python
### ZZ measurement
result = state_before_measurement

# Simplify the coefficent
norm_factor = 1/2
final_state = sp.simplify(result / norm_factor)

# Each index represents |q0 q1 q2⟩ → format: |q0 q1⟩ ⊗ |q2⟩
components = [sp.simplify(final_state[i]) for i in range(8)]
grouped = {
    "00": [components[0], components[1]],
    "01": [components[2], components[3]],
    "10": [components[4], components[5]],
    "11": [components[6], components[7]],
}

print("Apply gates to |ψ⟩ to get orginal input_q")
restoration_gate_map = {
    "00": I, 
    "01": X, 
    "10": Z, 
    "11": Z * X
    }
# |q0 q1⟩ = 00
trans_state_00 = restoration_gate_map["00"] * sp.Matrix(grouped["00"])
assert input_q.equals(trans_state_00), f"[ERROR] 00: {trans_state_00}"
print("[PASS] ZZ measurement for |q0 q1⟩ = 00")

# |q0 q1⟩ = 01
trans_state_01 = restoration_gate_map["01"] * sp.Matrix(grouped["01"])
assert input_q.equals(trans_state_01), f"[ERROR] 01: {trans_state_01}"
print("[PASS] ZZ measurement for |q0 q1⟩ = 01")

# |q0 q1⟩ = 10
trans_state_10 = restoration_gate_map["10"] * sp.Matrix(grouped["10"])
assert input_q.equals(trans_state_10), f"[ERROR] 10: {trans_state_10}"
print("[PASS] ZZ measurement for |q0 q1⟩ = 10")

# |q0 q1⟩ = 11
trans_state_11 = restoration_gate_map["11"] * sp.Matrix(grouped["11"])
assert input_q.equals(trans_state_11), f"[ERROR] 11: {trans_state_11}"
print("[PASS] ZZ measurement for |q0 q1⟩ = 11")
```
